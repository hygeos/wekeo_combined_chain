## Computed L3 FRP Visualisation

```python
"""
Plotting the gridded L3 FRP products. The variables available in the dataset are:
FRP_SWIR, FRP_SWIR_no_SAA, FRP_MWIR, and their respective mean, min, max, std, and count.
with derivatives for day and night.

like:
    day_FRP_SWIR_mean
    
NOTE: no_SAA could be systematically applied to all ?
"""

from wekeo_frp_l3 import plot_L3_FRP
from pathlib import Path
from wekeo_frp_l3 import config

Availables = [ 'FRP_SWIR', 'FRP_SWIR_no_SAA', 'FRP_MWIR' ]
var = 'FRP_MWIR'

colormap = "plasma"
figsize = (16 * 2, 10 * 2)

# Single variable
plot_L3_FRP.plot_L3_FRP(
    ds, 
    variable=f'day_{var}_mean',  
    cmap=colormap, 
    figsize=figsize,
    prefix="frp_slstr__",
    # save_fig_dir='./plots'
)
```
